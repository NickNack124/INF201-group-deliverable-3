## Task 1
The Updated "Character" class. 

```python
from character import Character

# The character class has 3 main attributes:
#   * name (str): The name of the character.
#   * hp_max (int > 0): Their max- and starting health points. (default: 100)
#   * damage (int >= 0): The default HP-amount they deal as damage. (default: 10)

# Let us initialize a character:
ch1 = Character("Johnathan")
ch1.introduce()
print(ch1)
ch1
```

```python
# The take_damage and heal methods modify the character's hp by a specified amount
ch1.take_damage(10)
ch1.heal(20)
```

```python
# We can also check whether or not the character is alive
print(f"{ch1.name} is alive? {ch1.is_alive}")
```

```python
# If we make a second character, we can have them attack
ch2 = Character("Dio", damage=200)
ch2.attack(ch1)
print(f"{ch1.name} is alive? {ch1.is_alive}")
```

```python
# Seems character 1 has died. That means the precious methods are unavailable to them.
ch1.attack(ch2)
ch2.attack(ch1)
ch1.heal(100)
```

There are more methods that can be performed, but they are mostly for the purpose of internal processing. These are the main methods the class is centered around.

## Task 3
The "Weapon" class in the "items" module.

```python
from items import Weapon

# The Weapon class has 2 main attributes:
#   * name (str): The name of the weapon
#   * damage_dice (Dice): The n-sided die rolled when calculating damage with the weapon

weapon = Weapon("Excalibur", damage_dice=d6)
print(weapon)
weapon
```

```python
# The weapons die can be rolled using roll_damage()

weapon.roll_damage()
ch1 = Character("Joseph", weapon=weapon)
# We decided to use our own damage formula for damage dealt when using a weapon.
# Instead of just doing the rolled damage, the damage it does is calculated as:
# self.damage + (0.1 * self.damage * the damage rolled by weapon)
# That way, stronger characters hit harder, even with a weapon
ch1.attack(ch2)
ch1.attack(ch2)
ch1.attack(ch2)
# (in the future this can be expanded to weapon types and proficiencies)
```

## Task 4
Subclasses for the "Character" class.

```python
from character import Mage, Warrior
from items import Armor

# The Warrior subclass has the same attributes as Character
# They also have a higher base damage and hp by default
ch3 = Warrior("Gunnar")
ch3.introduce()

# The Mage subclass has the same attributes as Character, plus:
#   * max_mana (int >= 0): The starting- and max amount of mana to be used for casting spells
#   * _mana (int >= 0): The mana resource itself
# We made the mage's base damage by default be lower, as well as their HP
ch4 = Mage("Wiz")
ch4.introduce()

# The Armor class has 2 main attributes:
#   * name (str): The name of the weapon
#   * armor (int >= 0): How much the armor protects from
chestplate = Armor("Holy steel", armor=12)
print(chestplate)
```

```python
# The armor can be worn by anyone in our version of the game, but the Warrior gets to have it by default
print(ch3)

# The mage has mana, which can be used to cast spells
print(ch4)
```

```python
ch3 = Warrior("Gunnar")
# Armor absorbs incoming damage, thus reducing damage received
ch3.take_damage(20)     # 20 becomes 15
ch4.take_damage(20)     # 20 remains 20

# The original request was to only let warrior have armor, so in order
# to compensate for the loss of uniqueness in the warrior-build, we also
# gave the subclass the ability to stall death in rare cases
ch3.take_damage(100)
# (you might need to re-run the script a few times to see him pull it off)
```

### Task 5
Simulating a battle with the "Battle" class.

```python
from battle import Battle

ch1_weapon = Weapon("Pluck", d6)
ch1 = Warrior("Johnathan", weapon=ch1_weapon)
ch2 = Mage("Dio", mana_max=50, damage=11)

# Battle has two main attributes:
# * character 1 (ch1): The first character in the battle
# * character 2 (ch2): The second character in the battle

# Simulating one round of battle between two characters
battle = Battle(ch1, ch2)
battle.play_round()
```

```python
# Now we can simulate a full battle between two characters
ch1.heal(100)
ch2.heal(100)
print(battle.simulate_battle())
```

```python
# We also have a failsafe for situations where battles drag on for too long (here: 30 rounds)
chara = Mage("Chara", armor=Armor("God's steel", 99999))
magoo = Warrior("Magoo", armor=Armor("Satan's sulfur", 99999))
battle = Battle(chara, magoo)
print(battle.simulate_battle())
```
